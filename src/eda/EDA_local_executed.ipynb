# Local exploratory data analysis
Executed on the bundled sample1 dataset without W&B. The original EDA.ipynb provides the artifact-backed workflow. Missing values are imputed inside the inference model. Prices are filtered to $10 through $350 inclusive.

```python
from pathlib import Path
import pandas as pd
root = Path.cwd()
if not (root / 'components').exists():
    root = root.parents[1]
df = pd.read_csv(root / 'components/get_data/data/sample1.csv')
print('Raw rows:', len(df))
df.info()
```

```python
print(df.describe().to_string())
print('Missing values:')
print(df.isna().sum().to_string())
print('Neighborhood counts:')
print(df.neighbourhood_group.value_counts().to_string())
print('First five rows:')
print(df.head().to_string(index=False))
```

```python
price_filtered = df[df.price.between(10, 350)].copy()
price_filtered['last_review'] = pd.to_datetime(price_filtered.last_review, errors='coerce')
print('Cleaned sample1 rows:', len(price_filtered))
print('Removed price outliers:', len(df) - len(price_filtered))
print('Review date dtype:', price_filtered.last_review.dtype)
print(price_filtered.price.describe().to_string())
```

## Findings
The sample has 20,000 rows and retains 19,001 after price filtering. Listing names, host names, review dates and reviews per month contain missing values. The review date is parsed as datetime. No missing values are imputed during cleaning; the complete inference pipeline handles missingness. Sample2 adds one out-of-boundary listing, documented in evidence/boundary_check.json.
